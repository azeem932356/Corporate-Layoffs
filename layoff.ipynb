```python
annual_layoffs = layoff.groupby('year')['total_laid_off'].sum().reset_index()
```

```python
annual_layoffs
```

```python
plt.figure(figsize=(5, 3))
plt.bar(annual_layoffs['year'], annual_layoffs['total_laid_off'], color='blue')
plt.title('Total Layoffs Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Layoffs')
plt.xticks(annual_layoffs['year'], rotation=0)
plt.show()
```

```python
company_layoffs = layoff.groupby('company')['total_laid_off'].sum().reset_index()
```

```python
company_layoffs = company_layoffs.sort_values(by='total_laid_off', ascending=False)
company_layoffs = company_layoffs.reset_index()
```

```python
company_layoffs = company_layoffs.drop(columns = ['index'])
```

```python
top_layoffs= company_layoffs.nlargest(15, 'total_laid_off')
```

```python
top_layoffs
```

```python
custom_colors = ['blue', '#66B3FF', '#99FF99', '#FFCC99', '#FFD700',
                 '#FF6347', '#ADFF2F', '#40E0D0', '#FF69B4', '#BA55D3']
```

```python
plt.figure(figsize=(10, 8))
plt.pie(top_layoffs['total_laid_off'], labels=top_layoffs['company'], autopct='%1.1f%%', colors=custom_colors)
plt.title('Top Companies by Total Layoffs')
plt.show()
```
